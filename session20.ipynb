```python
filepath="/content/ford_car_dataset.csv"
df=pd.read_csv(filepath)
```

```python
df
```

```python
# Separate Independent (X) and Dependent (Y) features
X = df.drop(columns=['price'])
Y = df['price']

# Print shapes
print("Shape of X (Independent Features):", X.shape)
print("Shape of Y (Dependent Feature):", Y.shape)
```

```python
# Identify categorical features in X
categorical_features = ['model', 'fuelType', 'transmission']

print("\nBefore One-Hot Encoding, X categorical columns:")
print(X[categorical_features].head(5))

# Perform one-hot encoding on the identified categorical columns in X
# Use drop_first=True to avoid multicollinearity
X_encoded = pd.get_dummies(X[categorical_features], drop_first=True, dtype=int)

# Drop the original categorical columns from X
X = X.drop(columns=categorical_features)

# Concatenate the processed X (numerical features) with the one-hot encoded features
X = pd.concat([X, X_encoded], axis=1)

print("\nAfter One-Hot Encoding, first 5 rows of X (showing new encoded columns):")
print(X.head(5))
print("\nShape of X after one-hot encoding:", X.shape)
```

```python
from sklearn.preprocessing import StandardScaler
numeric_feature=['year','mileage','tax','mpg','engineSize']
scaler=StandardScaler()
# Apply scaling to the numerical features in X, not df
X[numeric_feature]=scaler.fit_transform(X[numeric_feature])
print("\nFirst 5 rows of X after standard scaling of numeric features:")
print()
print(X[numeric_feature].head(5))
```

```python
from sklearn.model_selection import train_test_split
# Re-running this cell to ensure X_train and X_test are created from the fully preprocessed X.
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.33, random_state=42)
print("Shape of X_train (after preprocessing and splitting):", X_train.shape)
print("Shape of X_test (after preprocessing and splitting):", X_test.shape)
print("Shape of Y_train:", Y_train.shape)
print("Shape of Y_test:", Y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
# Re-fitting the model now that X_train and Y_train are correctly preprocessed.
model.fit(X_train, Y_train)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, Y_train)
y_pred=model.predict(X_test)
print("Intercept")
print(model.intercept_)
print()
print("Coefficients")
print(model.coef_)
```

```python
from sklearn.metrics import  r2_score

# Make predictions on the test set
print("\nPredicted Value:",y_pred[0: 11])
print("\nActual Value:",Y_test.tolist()[0: 11])

```

```python
from sklearn.metrics import  r2_score
r2=r2_score(Y_test,y_pred)
print("R2 Score:",r2)
```

```python
# The R² score of 0.84 indicates that the model explains 84% of the variance in car prices.
# This suggests the model is performing well, with predictions closely aligned to actual values,
# though some variability remains due to factors not included in the dataset.

# The R² score is 0.84, which means the model explains 84% of the variance in car prices.
# This indicates strong predictive performance — the model is capturing most of
# the important patterns.
# However, 16% of the variance remains unexplained,
# suggesting there are still factors not included in the dataset.
# Overall, the model is performing well and provides reliable predictions,
# though improvements are possible.
```
